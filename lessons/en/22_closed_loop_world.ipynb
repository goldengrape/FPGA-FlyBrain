# Lesson 22 | Give the fly a world: close the loop

A neural system becomes more interesting when its outputs change an environment that then produces the next inputs.

Today asks one question:

> **What makes an experiment closed-loop rather than a fixed replay?**

Primary new concept: **closed loop — output changes the future input.**

The toy world in this lesson is an engineering teaching model. It is **not a claim about real Drosophila behavior or MaleCNS sensory semantics.**

## 1. Concept ledger

**Already known:** host inputs, spike outputs, deterministic replay, connectome structure.

**New today:** **closed loop** — system output changes environment state, which changes later sensory input.

**Supporting roles:** a **sensory encoder** maps environment observations to neural stimulation; an **output decoder** maps selected neural activity to an action.

**Preview only:** biologically justified sensory mappings and descending-neuron behavior mappings.

## 2. Open-loop replay versus closed loop

In an open-loop replay, the input sequence is fixed before the run starts.

In a closed loop, the input at time (t+1) depends on what the system did at time (t).

That feedback is the key new idea. Encoder and decoder are explicit boundaries so assumptions remain visible.

## 3. Closed-loop picture

```mermaid
flowchart LR
  A["environment state"] --> B["sensory encoder"]
  B --> C["neural system"]
  C --> D["output decoder"]
  D --> E["action"]
  E --> A
```

## 4. Run: a deterministic one-dimensional toy world

Rules for this teaching world:

- target is fixed at position 3;
- if target is to the right, encoder emits `right`;
- the neural-system stub forwards that event to an output event;
- decoder turns `right` into action +1;
- environment updates the position.

These rules are intentionally simple and manually defined.

````python
position = 0
target = 3
trace = [position]

for step in range(5):
    if position < target:
        sensory_event = "right"
    elif position > target:
        sensory_event = "left"
    else:
        sensory_event = "at_target"

    output_event = sensory_event  # teaching stub, not a biological claim

    action = {
        "right": 1,
        "left": -1,
        "at_target": 0,
    }[output_event]

    position += action
    trace.append(position)
    print(step, sensory_event, action, position)

print("position trace:", trace)
````

## 5. Observe

Once the position reaches the target, the next observation changes to `at_target`, the action becomes zero, and future inputs differ from what they would have been under a fixed replay.

That dependence on prior output is the closed loop.

## 6. Engineering assumptions must be visible

The project documents require manual sensory/output mappings to be explicit.

In this toy world, all of these are assumptions:

- what the environment variable means;
- which sensory event represents “target is right”;
- which neural output maps to +1 motion;
- how action changes the world.

A future biological mapping needs its own evidence; it must not inherit these toy rules silently.

## 7. Deterministic replay still matters

Closed loop does not mean “unreproducible.”

If model version, network image, initial world state, mapping rules, input randomness, and PRNG seed are fixed, the complete trajectory should be replayable. That is the role of the full-system determinism requirement.
