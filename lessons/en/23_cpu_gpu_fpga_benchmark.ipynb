# Lesson 23 | Three machines, one experiment

CPU, GPU, and FPGA results are meaningful only when the experiment being compared is actually the same.

Today asks one question:

> **What must be held constant before a CPU/GPU/FPGA benchmark is fair?**

Primary new concept: **benchmark comparability — same model, same data, same input, defined metrics.**

## 1. Concept ledger

**Already known:** **Central Processing Unit (CPU)**, **Graphics Processing Unit (GPU)**, **Field-Programmable Gate Array (FPGA)**, latency, throughput, bandwidth, power as an engineering quantity, and reproducible replay.

**One primary concept today:** a **fair benchmark contract** — freeze model/data/input/measurement rules before comparing machines.

**Supporting metric:** energy per event simply combines already-known power and duration through `energy = power × time`, then divides by the work unit. It is not a second architecture concept.

**Preview only:** real measured CPU/GPU/FPGA implementations, board power instrumentation, and full P-001~P-008 reports.

## 2. A fast number without a workload is not a comparison

Before comparing machines, freeze at least:

- neuron-model semantics and numerical precision;
- exact network image/version;
- initial state and input events;
- warm-up and measurement window;
- what latency/throughput includes;
- power measurement method.

Changing the model while changing the machine makes the result ambiguous.

## 4. Run: compute metrics from synthetic measurements

The following numbers are **synthetic teaching measurements**. They are not claims about any real CPU, GPU, or FPGA.

```python
runs = [
    {"machine": "CPU-teaching", "events": 1_000_000, "seconds": 0.50, "watts": 40.0},
    {"machine": "GPU-teaching", "events": 1_000_000, "seconds": 0.20, "watts": 120.0},
    {"machine": "FPGA-teaching", "events": 1_000_000, "seconds": 0.25, "watts": 18.0},
]

for run in runs:
    throughput = run["events"] / run["seconds"]
    energy_j = run["watts"] * run["seconds"]
    energy_per_event = energy_j / run["events"]
    print(
        run["machine"],
        "events/s =", round(throughput),
        "J/event =", f"{energy_per_event:.8f}",
    )
```

## 5. Observe

Different metrics can disagree. A machine with the highest throughput need not have the lowest energy per event.

That is why the benchmark report should expose several metrics rather than compressing the experiment into one vague “faster” claim.

## 6. Latency and throughput answer different questions

- **latency**: how long one event/request/experiment takes;
- **throughput**: how much work completes per unit time;
- **energy per event**: how much measured energy is associated with one unit of work.

A benchmark must define its unit of work before these numbers are interpretable.

## 7. Power is especially easy to compare badly

Power can mean a tool estimate, board-level measurement, accelerator-only measurement, or whole-system wall power.

RMD-028 requires a documented method. Do not mix incomparable power boundaries and then call the ratio an architecture result.
