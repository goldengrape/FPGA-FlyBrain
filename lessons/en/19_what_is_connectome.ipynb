# Lesson 19 | What is a connectome?

So far we have built the machinery that can move spike events and synapse records. The next question is where a large real network comes from.

Today asks one question:

> **What information does a connectome give us, and what information does it not give us?**

Primary new concept: **connectome as a directed graph plus metadata.**

## 1. Concept ledger

**Already known:** neuron IDs, sparse synapse records, source-to-target edges, and event-driven updates.

**New today:** **connectome** — a dataset describing which neurons are connected to which other neurons, usually together with annotations or metadata.

**Preview only:** MaleCNS conversion, manifest/checksum, binary image layout, and full-scale loading.

## 2. A connectome is structure, not a running brain

For this course, represent a connectome as:

- **nodes**: neurons identified by stable IDs;
- **directed edges**: source neuron → target neuron connections;
- **edge metadata**: for example a weight or synapse count after an approved conversion rule;
- **node metadata**: annotations such as a cell type or region when available.

A connectome does **not** by itself tell us the current membrane voltage, current spike queue, random seed, or sensory input. Those belong to the running model or experiment state.

## 3. Small graph picture

```mermaid
flowchart LR
  N101["neuron 101"] --> N205["neuron 205"]
  N101 --> N330["neuron 330"]
  N205 --> N330
```

The arrow direction matters: source and target are not interchangeable.

## 4. Run: summarize a tiny connectome

The records below are deliberately tiny teaching data, not MaleCNS data.

````python
neurons = [
    {"id": 101, "type": "sensory-like"},
    {"id": 205, "type": "interneuron-like"},
    {"id": 330, "type": "output-like"},
]

edges = [
    (101, 205),
    (101, 330),
    (205, 330),
]

in_degree = {neuron["id"]: 0 for neuron in neurons}
out_degree = {neuron["id"]: 0 for neuron in neurons}

for source, target in edges:
    out_degree[source] += 1
    in_degree[target] += 1

print("neurons:", len(neurons))
print("directed edges:", len(edges))
print("in-degree:", in_degree)
print("out-degree:", out_degree)
````

## 5. Observe

Neuron 101 has two outgoing edges and no incoming edge in this tiny graph. Neuron 330 has two incoming edges and no outgoing edge.

The degree counts describe **graph structure**. They do not say whether any neuron is currently spiking.

## 6. IDs, metadata, and dynamic state are different things

A useful separation is:

- **neuron ID**: identifies an entity;
- **metadata**: describes that entity or connection;
- **model state**: changes while a simulation or FPGA run proceeds.

Mixing these categories makes reproducibility difficult. A converter should not silently invent dynamic membrane state while translating connectivity data.
