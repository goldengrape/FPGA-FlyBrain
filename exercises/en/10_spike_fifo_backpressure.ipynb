This is not only about first-in, first-out ordering. The key question is what happens to a new event when the queue is full: **who still owns it, whether it was actually accepted, and when it may be retried.**

## Objective and fixed contract

- fifo_push() returns (new_queue, accepted);
- when not full, append the new event;
- when full, accepted=False and the old queue remains unchanged;
- fifo_pop() removes the oldest event;
- empty pop returns None;
- a blocked event remains with the caller and may be retried after space opens.
