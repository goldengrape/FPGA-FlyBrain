## 本题目标与 fixed contract

- fifo_push() 返回 (new_queue, accepted)；
- 未满时，新 event 进入队尾；
- 已满时 accepted=False，而且旧 queue 完全不变；
- fifo_pop() 从队首取最早事件；
- empty pop 返回 None；
- blocked event 仍由 caller 保留，空间释放后可以重试同一个 event。
