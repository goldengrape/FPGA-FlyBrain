## 2. 先看整体结构

KV260 Starter Kit 不是“只有一颗 FPGA 的 PCB”。

你应该能区分：

- **K26 SOM**：包含 Zynq UltraScale+ MPSoC、DDR 等核心资源；
- **carrier card**：提供电源、JTAG/UART、microSD、Ethernet、Pmod、摄像头等物理连接；
- **thermal solution / fan**。

先找到 SOM，再找 carrier。不要从接口编号开始死记。

## 3. 必须能指出的接口

按 UG1089 的实物图，在自己的板上依次找到：

- **J12** — 12 V power input；
- **J4** — FTDI USB 2.0 UART + JTAG；
- **J3** — direct JTAG；
- **J11** — microSD；
- **J10** — 1 Gb/s Ethernet；
- **J2** — Pmod；
- **SW2** — SOM reset via POR_B；
- SOM **DS34** — PS done LED；
- SOM **DS35** — heartbeat LED；
- SOM **DS36** — PS status LED。

还要找到 carrier revision 标识。UG1089 明确存在 Rev. 1.0 与 Rev. 2.0，两者部分器件位置/功能不同。

## 4. 两个特别容易误解的点

### SW2 不是你的 `rst_n`

SW2 是 **SOM-level hard reset**。它会复位整个 SOM。它不会自动变成以后某个 neuron module 的 local `rst_n`。

### DS34 不是所有 programming 的“成功灯”

AMD 对 DS34 的定义是：**PS successfully loaded a PL design** 时点亮。

以后若使用 Vivado/JTAG direct program PL，不能仅凭 DS34 判断 direct programming 是否成功。
