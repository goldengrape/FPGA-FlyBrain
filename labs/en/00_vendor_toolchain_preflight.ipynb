# LAB-HW-00 — KV260 Vendor Toolchain Preflight

**One problem today: before connecting any FPGA board, prove that the development host has a usable AMD Vivado installation, JTAG driver, and KV260 board definition.**

Do not connect the board, write RTL, or generate a bitstream. Keeping tool setup separate makes later target-detection failures attributable.

**Project Trace:** RMD-012 · T-HW-001/T-HW-011

## 1. What you need

- the computer that will be the **development host**;
- enough disk space for AMD Vivado;
- network access for the installer and Vivado Board Store;
- **no powered KV260 and no microSD today.**

### Course authoring baseline

This first lab batch uses **Vivado 2026.1** as its authoring baseline, grounded in AMD's current 2026.1 Board Flow documentation and current KV260 Board Store data.

This is not a claim that a physical board has passed. Real target discovery becomes T-HW-002 in LAB-HW-02.

## 2. Installation requirement

Install AMD Vivado 2026.1 with device support for **Zynq UltraScale+ MPSoC / Kria K26**.

Then run:

```bash
vivado -version
```

If the shell cannot find `vivado`, stop here and fix the installation/environment.

Do not substitute OSS CAD Suite/Yosys for this step. Lesson 13 generic synthesis and KV260 target-specific implementation/programming are different tool layers.

## 3. Install/refresh KV260 board definitions

In Vivado:

1. **Tools → Vivado Store**;
2. open **Boards**;
3. Refresh;
4. search for Kria / KV260;
5. install the AMD/Xilinx KV260/K26 SOM and companion-card board data.

The course intentionally checks for current `*kv260*` board parts instead of hard-coding an obsolete board-part revision copied from an old tutorial.

## 4. Run the course preflight

From the repository root:

```bash
vivado -mode batch -nolog -nojournal \
  -source boards/kv260/scripts/check_vivado.tcl \
  | tee lab-hw-00-vivado-preflight.txt
```

Expected evidence:

- Vivado version;
- at least one KV260-related board part.

A non-zero exit means LAB-HW-00 has **not** passed.

## 5. Save Evidence

Retain:

- `lab-hw-00-vivado-preflight.txt`;
- OS/version;
- Vivado version;
- KV260 board-part list;
- current Git commit.

A screenshot may supplement the text log, but must not replace it.

## 6. If it does not work

**A. `vivado` is not found**  
→ fix installation/PATH before touching hardware.

**B. Vivado runs but reports `NO_KV260_BOARD_PART`**  
→ Vivado Store → Boards → Refresh → install KV260/K26 board data, then rerun.

**C. Your Vivado version is not 2026.1**  
→ record it and stop the physical track. Changing the course baseline is a specification decision, not a silent per-student substitution.
