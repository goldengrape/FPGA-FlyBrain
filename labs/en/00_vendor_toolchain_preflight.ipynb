# LAB-HW-00 — KV260 Vendor Toolchain Preflight

**One problem today: before connecting any FPGA board, prove that the development host has a usable AMD Vivado installation, JTAG driver, and KV260 board definition.**

Do not connect the board, write RTL, or generate a bitstream. Keeping tool setup separate makes later target-detection failures attributable.

**Project Trace:** RMD-012 · T-HW-001/T-HW-011

## 1. What you need

- an **x86-64** development host;
- AMD Vivado 2026.1 installation media;
- sufficient disk space;
- network access for the installer and Vivado Board Store;
- **no powered KV260 and no microSD today.**

### First-release host-OS policy

To keep the zero-experience path reproducible, the first course release keeps its documented beginner path to **native Windows / Linux** hosts that AMD supports for Vivado 2026.1. Prefer:

- an AMD-supported Windows 10/11 release;
- an AMD-supported Ubuntu 22.04 / 24.04 release.

macOS, WSL, and virtual-machine USB/JTAG passthrough are not part of the first guaranteed beginner path. They may later receive porting notes, but must not silently replace the tested path.

### Course authoring baseline

This batch selects **Vivado 2026.1** as its **authoring candidate baseline** so prose, scripts, and screenshots do not drift. It is not yet a physically validated supported baseline; that status requires a real-KV260 dry run with retained evidence. Real target discovery does not become T-HW-002 evidence until LAB-HW-02.

## 2. Install Vivado and the cable driver

Install AMD Vivado 2026.1 with **Zynq UltraScale+ MPSoC / Kria K26** device support.

Then run:

```bash
vivado -version
```

If the shell cannot find `vivado`, stop here and fix the installation/environment.

### JTAG cable driver

**Windows:** select **Install Cable Drivers** in the Vivado installer. For post-install repair/installation, AMD UG973 gives this Administrator-command-prompt flow (replace `%VIVADO_INSTALL_DIR%` if your installation lives elsewhere):

```bat
cd %VIVADO_INSTALL_DIR%\data\xicom\cable_drivers\nt64
install_drivers_wrapper.bat %TEMP%
```

**Linux:** after installing Vivado, AMD requires the cable-driver installer to be run with root/sudo privileges:

```bash
cd <Vivado Install Dir>/data/xicom/cable_drivers/lin64/install_script/install_drivers/
sudo ./install_drivers
```

LAB-HW-00 can record that the driver-install step completed, but **LAB-HW-02 real target discovery is what proves the cable/JTAG path actually works**. Do not turn “installer succeeded” into “board verified.”

Do not substitute OSS CAD Suite/Yosys for Vivado here. Lesson 13 generic synthesis and KV260 target-specific implementation/programming are different tool layers.

## 3. Install/refresh KV260 board definitions

In Vivado:

1. **Tools → Vivado Store**;
2. open **Boards**;
3. Refresh;
4. search for Kria / KV260;
5. install the AMD/Xilinx KV260/K26 SOM and companion-card board data.

The course intentionally checks for current `*kv260*` board parts instead of hard-coding an obsolete board-part revision copied from an old tutorial.

## 4. Run the course preflight

From the repository root:

```bash
vivado -mode batch -nojournal \
  -log lab-hw-00-vivado-preflight.log \
  -source boards/kv260/scripts/check_vivado.tcl
```

Expected evidence:

- Vivado version;
- at least one KV260-related board part.

A non-zero exit means LAB-HW-00 has **not** passed.

## 6. If it does not work

**A. `vivado` is not found**  
→ fix installation/PATH before touching hardware.

**B. Vivado runs but reports `NO_KV260_BOARD_PART`**  
→ Vivado Store → Boards → Refresh → install KV260/K26 board data, then rerun.

**C. Your Vivado version is not 2026.1**  
→ record it and stop the physical track. Changing the course baseline is a specification decision, not a silent per-student substitution.
