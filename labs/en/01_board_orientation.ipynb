Find the SOM and carrier first. Then identify connector names.

## 3. Interfaces you must locate

Using the UG1089 board figure, locate on your actual board:

- **J12** — 12 V power input;
- **J4** — FTDI USB 2.0 UART + JTAG;
- **J3** — direct JTAG;
- **J11** — microSD;
- **J10** — 1 Gb/s Ethernet;
- **J2** — Pmod;
- **SW2** — SOM reset through POR_B;
- SOM **DS34** — PS done LED;
- SOM **DS35** — heartbeat LED;
- SOM **DS36** — PS status LED.

Also locate the carrier revision marking. UG1089 documents both Rev. 1.0 and Rev. 2.0, with some revision-specific differences.

## 4. Two common misconceptions

### SW2 is not your module `rst_n`

SW2 is a **SOM-level hard reset**. It does not automatically become the local reset of a later neuron RTL block.

### DS34 is not a universal programming-success LED

AMD defines DS34 as lit when the **PS successfully loaded a PL design**.

If a later lab directly programs PL through Vivado/JTAG, DS34 alone is not the oracle for that direct programming path.
